```python
import pandas as pd
import matplotlib.pyplot as plt

df_inquiries = pd.read_csv(r"c:\Users\SDN PESISIR\Desktop\Data Analyst\Dataset Griyalaju Marketplace\dataset\inquiries.csv")
df_agencies = pd.read_csv(r"c:\Users\SDN PESISIR\Desktop\Data Analyst\Dataset Griyalaju Marketplace\dataset\agencies.csv")
df_agents = pd.read_csv(r"c:\Users\SDN PESISIR\Desktop\Data Analyst\Dataset Griyalaju Marketplace\dataset\agents.csv")
df_listings = pd.read_csv(r"c:\Users\SDN PESISIR\Desktop\Data Analyst\Dataset Griyalaju Marketplace\dataset\listings.csv")
df_transactions = pd.read_csv(r"c:\Users\SDN PESISIR\Desktop\Data Analyst\Dataset Griyalaju Marketplace\dataset\transactions.csv")

print(df_agencies.info())
print("*"*100)
print(df_agents.info())
print("*"*100)
print(df_inquiries.info())
print("*"*100)
print(df_listings.info())
print("*"*100)
print(df_transactions.info())
```

```python
df_inquiries = df_agencies.dropna(how='all')
df_agencies = df_agencies.dropna(how='all')
df_agents = df_agents.dropna(how='all')
df_listings = df_listings.dropna(how='all')
df_transactions = df_transactions.dropna(how='all')
```

```python
df_transactions['transaction_date'] = pd.to_datetime(df_transactions['transaction_date'])
df_transactions['year'] = df_transactions['transaction_date'].dt.year


plt.plot(df_transactions.groupby('year')['sale_price_idr'].sum())
plt.title('Jumlah total pendapatan berdasarkan tahun')
plt.show()
```

```python
df_counts_city = df_listings.groupby(['city', 'status']).size().unstack(fill_value=0)
df_counts_city.plot(kind='bar', figsize=(10, 6))

plt.title('Jumlah Status Berdasarkan Kota')
plt.xlabel('Kota')
plt.ylabel('Jumlah Frekuensi')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```

```python
df_counts_province = df_listings.groupby(['province', 'status']).size().unstack(fill_value=0)
df_counts_province.plot(kind='bar', figsize=(10, 6))

plt.title('Jumlah Status Berdasarkan Provinsi')
plt.xlabel('Kota')
plt.ylabel('Jumlah Frekuensi')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```

```python
df_listings['count_property'] = df_listings['property_type']
df_count_property = df_listings.groupby(['count_property', 'property_type']).size().unstack(fill_value=0)

df_count_property.plot(kind='bar', figsize=(10, 6))

plt.title('Jumlah Properti Berdasarkan jenis properti')
plt.xlabel('Properti')
plt.ylabel('Jumlah Frekuensi')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```
